```python
import numpy as np
import pandas as pd
import scipy.stats as stats
```

## Random Distributions

```python
B0_log_rand = np.random.normal(13.25,0.75,int(1e+05))
v_m_rand = stats.maxwell.rvs(scale = 26500000,size=int(1e+05))
P0_log_rand = np.random.normal(-0.6,0.3,int(1e+05))

t_obvs_rand = np.random.uniform(0,(10**9)*(3.154e+07),int(1e+05))
```

```python
B0_rand = 10**B0_log_rand
P0_rand = 10**P0_log_rand
```

```python
TotalInput = pd.DataFrame({'B0': B0_rand[:], 'v_m': v_m_rand[:], 'P0': P0_rand[:], 't_obvs': t_obvs_rand[:]})
TotalInput
```

## Slicing Data
